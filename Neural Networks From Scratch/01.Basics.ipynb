```python
import numpy as np
```

Coding A single Neuron

![Tokenization](https://raw.githubusercontent.com/bhattacharjeeajay12/hands-on-ai-ml/main/content/images/nnfs02.png)

```python
inputs = [1.0, 2.0, 3.0, 2.5]
weights = [0.2, 0.8, -0.5, 1.0]
biases = 2.0

output = (
    inputs[0] * weights[0]
    + inputs[1] * weights[1]
    + inputs[2] * weights[2]
    + inputs[3] * weights[3]
    + biases
)
print(output)
```

Coding A single Layer of 3 neuron (fully connected)
![Tokenization](https://raw.githubusercontent.com/bhattacharjeeajay12/hands-on-ai-ml/main/content/images/nnfs01.png)

```python
inputs = [1.0, 2.0, 3.0, 2.5]
weights1 = [0.2, 0.8, -0.5, 1.0]
weights2 = [0.5, -0.91, 0.26, -0.5]
weights3 = [-0.26, -0.27, 0.17, 0.87]

biase1 = 2.0
biase2 = 3.0
biase3 = 0.5

outputs = [inputs[0]*weights1[0] + inputs[1]*weights1[1] + inputs[2]*weights1[2] + inputs[3]*weights1[3] + biase1,
           inputs[0]*weights2[0] + inputs[1]*weights2[1] + inputs[2]*weights2[2] + inputs[3]*weights2[3] + biase2,
           inputs[0]*weights3[0] + inputs[1]*weights3[1] + inputs[2]*weights3[2] + inputs[3]*weights3[3] + biase3]

print(outputs)
```

```python
inputs = [1.0, 2.0, 3.0, 2.5]
weights = [[0.2, 0.8, -0.5, 1.0],
           [0.5, -0.91, 0.26, -0.5],
           [-0.26, -0.27, 0.17, 0.87]]
biases = [2.0, 3.0, 0.5]

layer_outputs = np.dot(inputs, np.array(weights).T) + biases # it is WtX.
print(layer_outputs)
```
